```python
import random

roll_number = 1024170097

random.seed(roll_number)

numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)

odd_numbers = [num for num in numbers if num % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Number of odd numbers:", len(odd_numbers))

even_numbers = [num for num in numbers if num % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Number of even numbers:", len(even_numbers))

def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True

prime_numbers = [num for num in numbers if is_prime(num)]

print("\nPrime numbers:")
print(prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

most_frequent = max(frequency, key=frequency.get)
most_frequent_count = frequency[most_frequent]

print("\nMost frequent number:", most_frequent)
print("Number of occurrences:", most_frequent_count)
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 0]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

union = A.union(B)
print("Union:", union)

intersection = A.intersection(B)
print("Intersection:", intersection)

A_difference_B = A.difference(B)
B_difference_A = B.difference(A)

print("A - B:", A_difference_B)
print("B - A:", B_difference_A)

symmetric_difference = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_difference)

print("Is A subset of B:", A.issubset(B))
print("Is B superset of A:", B.issuperset(A))

X = int(input("Enter a value X: "))
A.discard(X)

print("Set A after discard:", A)
```

```python
my_dict = {
    "name": "Jaskaran Singh Bedi",
    "roll_no": "1024170097",
    "branch": "Computer Science",
    "age": 20,
    "city": "Ambala Cantt"
}

# i. Rename city to location
my_dict["location"] = my_dict.pop("city")
print("After renaming city:", my_dict)

# ii. Add CGPA
my_dict["cgpa"] = 9.7
print("After adding CGPA:", my_dict)

# iii. Increase age by 1
my_dict["age"] += 1
print("After updating age:", my_dict)

# iv. Delete branch using pop()
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print("Using pop():", dict_pop)

# Delete branch using del
dict_del = my_dict.copy()
del dict_del["branch"]
print("Using del:", dict_del)

# pop() removes the key and returns its value, while del only removes the key.

# v. Iterate using items()
for key, value in my_dict.items():
    print(key, "→", value)

# vi. Check whether email exists
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present in the dictionary.")

# vii. Create friend's dictionary
friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024170100",
    "branch": "Electronics",
    "age": 21,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:", merged_dict)

# When both dictionaries have the same key, the value from friend_dict wins
# because it appears second in {**my_dict, **friend_dict}.

# viii. Dictionary comprehension for string values
string_dict = {
    key: value for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Dictionary containing only string values:", string_dict)
```
