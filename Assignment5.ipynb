Q2 B i ARRAY X

```python

import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

max_count = counts.max()
modes = values[counts == max_count]

print("Array:", x)
print("Most frequent value(s):", modes)
print("Maximum frequency:", max_count)

for mode in modes:
    print(f"Indices of {mode}:", np.where(x == mode)[0])
```

Q4

```python

import numpy as np
Jaskaran = np.linspace(10, 100, 25)

print("Array:", Jaskaran)
print("Dimensions:", Jaskaran.ndim)
print("Shape:", Jaskaran.shape)
print("Total elements:", Jaskaran.size)
print("Data type:", Jaskaran.dtype)
print("Memory consumed (bytes):", Jaskaran.nbytes)

print("Transpose using T:", Jaskaran.T)
row_array = Jaskaran.reshape(1, 25)
print("Row array shape:", row_array.shape)

transposed_array = row_array.T
print("Transposed array shape:", transposed_array.shape)
print("Transposed array:\n", transposed_array)
```

```python

import numpy as np

ucs420_Jaskaran = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:\n", ucs420_Jaskaran)

print("Mean:", np.mean(ucs420_Jaskaran))
print("Median:", np.median(ucs420_Jaskaran))
print("Maximum:", np.max(ucs420_Jaskaran))
print("Minimum:", np.min(ucs420_Jaskaran))
print("Unique elements:", np.unique(ucs420_Jaskaran))

reshaped_ucs420_Jaskaran = ucs420_Jaskaran.reshape(4, 3)
print("\nReshaped Array (4 x 3):\n", reshaped_ucs420_Jaskaran)

resized_ucs420_Jaskaran = np.resize(ucs420_Jaskaran, (2, 3))
print("\nResized Array (2 x 3):\n", resized_ucs420_Jaskaran)
```
